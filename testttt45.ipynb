```python
import matplotlib.pyplot as plt
import pandas as pd

# Create the DataFrame
data = {
    "Shift Name": ["General", "Morning (400)", "Evening (400)", "Late Night (400)"],
    "Login (IN)": ["9:00 a.m.", "4:59 a.m.", "3:30 p.m.", "8:00 p.m."],
    "Logout (OUT)": ["6:00 p.m.", "1:59 p.m.", "12:30 a.m.", "5:00 a.m."],
    "Allowance (Rs/day)": [0, 400, 400, 400]
}

df = pd.DataFrame(data)

# Plot table
fig, ax = plt.subplots(figsize=(8, 2))
ax.axis('off')
table = ax.table(cellText=df.values, colLabels=df.columns, cellLoc='center', loc='center')
table.auto_set_font_size(False)
table.set_fontsize(12)
table.scale(1, 2)

# Save the figure as PNG
plt.savefig("shift_table.png", bbox_inches='tight', dpi=300)
```

```python
def total_vowels(input_string):
    vowels_list = ['a', 'e', 'i', 'o', 'u']
    count = 0

    if len(input_string) == 0:
        return "Empty string"
    
    input_string = input_string.lower()

    for i in input_string:
        if i in vowels_list:
            count += 1
    return count
    
```

```python
intput_str = "Rakesh"
print("Total no of vowel in given string",total_vowels(intput_str))
```

```python
#🔹 2. Start Spark Session with Snowflake Connector

from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("PySpark to Snowflake") \
    .config("spark.jars", "/path/to/spark-snowflake_2.12-<version>.jar,/path/to/snowflake-jdbc-<version>.jar") \
    .getOrCreate()
```

```python
# Snowflake connection options
sfOptions = {
    "sfURL": "https://MQGDRPX-CEC82858.snowflakecomputing.com",
    "sfUser": "nikkygupta",
    "sfPassword": "Nikkygupta@12345",
    "sfDatabase": "dbnikky9july",
    "sfSchema": "schnikky9july",
    "sfWarehouse": "whnikky9july",
    "sfRole": "ACCOUNTADMIN",
    "dbtable": table_name
}
```

```python
df_from_sf = spark.read \
  .format("snowflake") \
  .options(**sfOptions) \
  .option("dbtable", "customer_table_nik9july") \
  .load()

df_from_sf.show()
```
